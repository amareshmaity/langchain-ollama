### Multiturn conversation


```python
## Turn 1
message_history = [
    SystemMessage(content="You are an expert python tutor. Answer preciously"),
    HumanMessage(content="What is eval in python?")
]

ai_message = llm.invoke(message_history)
message_history.append(ai_message)
display_markdown_text(ai_message.content)
```

```python
## Turn 2
message_history.append(HumanMessage(content="Give some code example so that eval can be easily understood"))
ai_message = llm.invoke(message_history)
message_history.append(ai_message)
display_markdown_text(ai_message.content)
```

### Parameters inside model

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from IPython.display import display, clear_output
import time

prompts = ChatPromptTemplate.from_template(template="Answer the user's question precisely for topic: {topic}")

llm = ChatOllama(
    model="deepseek-coder:latest",
    temperature=0.3,
    stream=True,
    max_token=500
)

parser = StrOutputParser()

chain = prompts | llm | parser

streamed = chain.stream({"topic": "what is custom exception handling in python"})

print(f"...thinking...")

partial_message = ""
display_handle = display(Markdown(""), display_id=True)

for chunk in streamed:
    partial_message += chunk
    
    print(chunk, flush=True, end="")
    time.sleep(0.03)

clear_output(wait=True)
display_markdown_text(partial_message)
```

### PromptTemplate in chain

```python
prompt = PromptTemplate.from_template(
    "You are an {role}, write a poem about {topic} in {language} language under 12 lines and 3 stanger"
)
llm = ChatOllama(model="deepseek-r1:1.5b", temperature=0.7)
parser = StrOutputParser()

chain = prompt | llm | parser

response = chain.invoke({"role": "poet", "topic": "blackhole", "language": "english"})

display_markdown_text(response)
```

### partial method

```python
from urllib import response


template = PromptTemplate(
    input_variables=["role", "tone"],
    template="You are a {role}. Answer the user's question in {tone}. {user_question}"
)
template = template.partial(tone="deterministic")

llm = ChatOllama(model="deepseek-r1:1.5b", temperature=0.3)
chain = template | llm | parser

response = chain.invoke({"role": "mathematics expert", "user_question": "Alex starts with $60. He buys 3 notebooks for $7 each. He also buys a water bottle for $14. The store applies a 10% discount to his total purchase before tax. How much change does Alex get back from his $60? Show your step-by-step reasoning."})

display_markdown_text(response)
```
