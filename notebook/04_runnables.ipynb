## Runnable

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

### Runnable Interface
- .invoke()
- .stream()
- .batch()
- .ainvoke()
- .astream()
- .abatch()

```python
## .invoke()

prompts = ChatPromptTemplate.from_messages([
    ("system", "write short description with 4 to 5 lines about given topic"),
    ("human", "{topic}")
])

llm = ChatOllama(
    model="deepseek-r1:1.5b",
    temperature=0.3
)

parser = StrOutputParser()

chain = prompts | llm | parser

chain.invoke({"topic": "langchain"})
```

```python
## .stream
import time
response = chain.stream({"topic": "langchain"})
for chunk in response:
    print(chunk, flush=True, end="")
    time.sleep(0.3)
```

```python
res = llm.invoke("What is blackhole")
print(res.content)
```

```python
## .batch()
input = [
    {"topic": "langchain"},
    {"topic": "langflow"},
    {"topic": "langsmith"},
]

responses = chain.batch(inputs=input)
```

```python
type(responses)
```

```python
topic = [dic["topic"] for dic in input]
topic
```

```python
for topic, response in zip(topic, responses):
    print()
    print(f"{topic}: {response}")
    print()
```
