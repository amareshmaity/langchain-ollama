### Runnable Interface
- .invoke()
- .stream()
- .batch()
- .ainvoke()
- .astream()
- .abatch()

```python
## .invoke()

prompts = ChatPromptTemplate.from_messages([
    ("system", "write short description with 4 to 5 lines about given topic"),
    ("human", "{topic}")
])

llm = ChatOllama(
    model="deepseek-r1:1.5b",
    temperature=0.3
)

parser = StrOutputParser()

chain = prompts | llm | parser

chain.invoke({"topic": "langchain"})
```

```python
## .stream
import time
response = chain.stream({"topic": "langchain"})
for chunk in response:
    print(chunk, flush=True, end="")
    time.sleep(0.3)
```

```python
## .ainvoke()
import asyncio
import time
from langchain_ollama import ChatOllama

# A simple background task that prints a message every second
async def background_counter():
    for i in range(1, 10):
        await asyncio.sleep(1)
        print(f"   [Background Task]: Working for {i} seconds...")

async def chat(input_text): 
    llm = ChatOllama(model="deepseek-r1:1.5b") 
    print("[LLM]: Request sent to DeepSeek...") 
    
    start_time = time.time()
    response = await llm.ainvoke(input_text) 
    end_time = time.time()
    
    print(f"\n[LLM]: Execution successful in {end_time - start_time:.2f} seconds!")
    print(f"[LLM Response]: {response.content.strip()}")

# Use asyncio.gather to run both functions at the exact same time
print("Starting concurrent execution proof...\n")
await asyncio.gather(
    chat("Write a medium-length essay on the future of AI."),
    background_counter()
)
```

## RunnableParallel

```python
from langchain_core.runnables import RunnableParallel
```

```python
## Two call simultaneously
llm = ChatOllama(model="deepseek-r1:1.5b")

pros_chain = (
    ChatPromptTemplate.from_template("Write pros of {topic}")
    | llm
    | StrOutputParser()
)

cons_chain = (
    ChatPromptTemplate.from_template("Write cons of {topic}")
    | llm
    | StrOutputParser()
)

parallel_chain = RunnableParallel({"pros_chain": pros_chain, "cons_chain": cons_chain})

parallel_chain.invoke({"topic": "python"})
```

```python
from langchain_community.vectorstores import FAISS
```

```python
from langchain_ollama import OllamaEmbeddings
from langchain_core.runnables import RunnablePassthrough
```

```python
embedder = OllamaEmbeddings(model="mxbai-embed-large:latest")
docs = ["Langchain is framwork to build llm appliction", "langchain uses '|' sysmbol to build chain with components"]
vector_store = FAISS.from_texts(docs, embedder)
retriever = vector_store.as_retriever()
```

```python
setup = RunnableParallel({
    "context": retriever,
    "question": RunnablePassthrough()
})

prompts = ChatPromptTemplate.from_messages([
    ("system", "Use below context to answer user query \n {context}"),
    ("human", "{question}")
])

def format_docs(docs):
    return "\n\n".join(d.page_content for d in docs)

chain = (
    setup
    | {"context": lambda x: format_docs(x["context"]), "question": lambda x: x["question"]}
    | prompts
    | llm
    | StrOutputParser()
)

response = chain.invoke("tell me about langchain")
```

### Multiperspective Analysis

```python
analysis_chain = RunnableParallel({
    "definition": (
        ChatPromptTemplate.from_template("Write the definition of {topic} within 20 words")
        | llm
        | StrOutputParser()
    ),

    "description": (
        ChatPromptTemplate.from_template("Write short description about {topic}. Give the overview of the {topic}")
        | llm
        | StrOutputParser()
    ),

    "pros": (
        ChatPromptTemplate.from_template("Write pros of {topic}.")
        | llm
        | StrOutputParser()
    ),

    "cons": (
        ChatPromptTemplate.from_template("Write cons of {topic}.")
        | llm
        | StrOutputParser()
    ),

    "future": (
        ChatPromptTemplate.from_template("Write about the furure of {topic}.")
        | llm
        | StrOutputParser()
    )

})

analysis_chain.invoke({"topic": "javascript language"})
```

### Performance time: sequential VS parallel llm call

```python
from langchain_ollama import ChatOllama
from langchain_core.output_parsers import StrOutputParser
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOllama(model="deepseek-r1:1.5b")

pros_chain = (
    ChatPromptTemplate.from_template("pros of {topic}")
    | llm
    | StrOutputParser()
)
cons_chain = (
    ChatPromptTemplate.from_template("cons of {topic}")
    | llm
    | StrOutputParser()
)
```

```python
import time
from langchain_core.runnables import RunnableParallel

# Sequential — each runs after the previous
start = time.time()
pros    = pros_chain.invoke({"topic": "LangChain"})
cons    = cons_chain.invoke({"topic": "LangChain"})
print(f"Sequential: {time.time() - start:.2f}s")

# Parallel — all run at the same time
start = time.time()
result = RunnableParallel({"pros": pros_chain, "cons": cons_chain}).invoke({"topic": "LangChain"})
print(f"Parallel:   {time.time() - start:.2f}s")
# Parallel ~= time of slowest single chain (not sum of all!)
```
