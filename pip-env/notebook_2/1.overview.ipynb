## Langchain version 2026-september

LangChain is an open-source development framework designed to build applications powered by large language models (LLMs).

Agent = Model + Harness

The harness is everything around the model loop: the prompt, the tools, and any middleware that shapes behavior.

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
```

```python
# Calculation tools
def sum(a, b):
    """Get sum of two numbers"""
    return a+b


# Agent
agent = create_agent(
    model="ollama:llama3.2:1b",
    tools=[sum],
    system_prompt="You are an helpful assistant"
)

result = agent.invoke(
    input={"messages": [{"role": "user", "content": "what is the sum of 3 and 4"}]}
) 
```
